```python
df = pd.read_csv("time_series_60min_singleindex.csv",
                 sep=",",
                 header=0,
                 low_memory=False)

print("Zeilen:", len(df))
print("Spalten:", len(df.columns))
print(df.head())
```

```python
# Nur Deutschland-Spalten auswählen
de_columns = ["utc_timestamp",
              "DE_load_actual_entsoe_transparency",
              "DE_load_forecast_entsoe_transparency", 
              "DE_solar_generation_actual",
              "DE_wind_onshore_generation_actual"]

df_de = df[de_columns]
print(df_de.head(10))
print("\nFehlende Werte pro Spalte:")
print(df_de.isnull().sum())
```

```python
# TRANSFORM
```

```python
# Schritt 1: Spalten umbenennen (lesbarer)
df_de = df_de.rename(columns={
    "utc_timestamp": "timestamp",
    "DE_load_actual_entsoe_transparency": "load_actual_mw",
    "DE_load_forecast_entsoe_transparency": "load_forecast_mw",
    "DE_solar_generation_actual": "solar_mw",
    "DE_wind_onshore_generation_actual": "wind_mw"
})

# Schritt 2: Solar NaN mit 0 ersetzen (nachts = kein Solar)
df_de["solar_mw"] = df_de["solar_mw"].fillna(0)

# Schritt 3: Restliche NaN Zeilen entfernen
df_de = df_de.dropna()

print("Zeilen nach Bereinigung:", len(df_de))
print("\nFehlende Werte nach Bereinigung:")
print(df_de.isnull().sum())
```

```python
# Neue Spalte: Gesamte erneuerbare Energie
df_de["renewables_mw"] = df_de["solar_mw"] + df_de["wind_mw"]

# Neue Spalte: Anteil Erneuerbarer am Verbrauch (in Prozent)
df_de["renewables_share_pct"] = (df_de["renewables_mw"] / df_de["load_actual_mw"] * 100).round(1)

# Neue Spalte: Vorhersagefehler (wie gut war die Prognose?)
df_de["forecast_error_mw"] = df_de["load_actual_mw"] - df_de["load_forecast_mw"]

print(df_de[["timestamp", "load_actual_mw", "renewables_mw", "renewables_share_pct", "forecast_error_mw"]].head(10))
```

```python
# Timestamp in echtes Datum umwandeln
df_de["timestamp"] = pd.to_datetime(df_de["timestamp"])

# Neue Spalten aus dem Datum ableiten
df_de["year"] = df_de["timestamp"].dt.year
df_de["month"] = df_de["timestamp"].dt.month
df_de["hour"] = df_de["timestamp"].dt.hour
df_de["weekday"] = df_de["timestamp"].dt.day_name()

print(df_de[["timestamp", "year", "month", "hour", "weekday"]].head(5))
print("\nDatentyp timestamp:", df_de["timestamp"].dtype)
```

```python
import sqlite3

# Verbindung zur Datenbank herstellen
conn = sqlite3.connect("energy_data.db")

# Daten in Datenbank speichern
df_de.to_sql("energy_de", 
             conn, 
             if_exists="replace",
             index=False)

print("Gespeichert!")
print("Anzahl Zeilen in DB:")
result = conn.execute("SELECT COUNT(*) FROM energy_de").fetchone()
print(result[0])

conn.close()
```

```python
conn = sqlite3.connect("energy_data.db")

# Erste 5 Zeilen aus der Datenbank lesen
result = pd.read_sql("SELECT * FROM energy_de LIMIT 5", conn)
print(result)

# Durchschnittswerte pro Jahr
yearly = pd.read_sql("""
    SELECT year,
           ROUND(AVG(load_actual_mw), 0) as avg_load_mw,
           ROUND(AVG(renewables_share_pct), 1) as avg_renewables_pct
    FROM energy_de
    GROUP BY year
    ORDER BY year
""", conn)
print("\nDurchschnitt pro Jahr:")
print(yearly)

conn.close()
```
